```python
#Install SAMtools
print("Installing SAMtools...")
!apt-get -y install samtools
```

```python
#Download Example BAM File (1000 Genomes - Chromosome 20)
print("Downloading example BAM file...")
!wget -o example.bam \ https://ftp.1000genomes.ebi.ac.uk/vol1/ftp/phase3/data/HG00096/alignment/HG00096.chrom20.ILLUMINA.bwa.GBR.low_coverage.20120522.

!wget -o example.bam.bai\ https://ftp.1000genomes.ebi.ac.uk/vol1/ftp/phase3/data/HG00096/alignment/HG00096.chrom20.ILLUMINA.bwa.GBR.low_coverage.20120522.
#
```

```python
#check File Sizes
print("\nChecking download files:")
!ls -lh example.bam
```

```python
#Sort BAM file
print("\nSorting BAM file...")
!samtools sort -o example.sorted.bam example.bam
```

```python
#Index Sorted BAM
print("Indexing sorted BAM...")
!samtools index example.sorted.bam
```

```python
#Filter reads with Mapping Quality >=30
print("Filtering reads with MAPQ >= 30...")
!samtools view -b -q 30 -o example.filtered.bam example.sorted.bam
```

```python
#Remove Unmapped Reads
print("Removing unmapped reads...")
!samtools view -b -F 4 -o example.mapped.bam example.sorted.bam 20:1-50000
```

```python
#Extract Specific Genomic Region (Chr20:1-50000)
print("Extracting specific genomic region (Chr20:1-50000)...")
!samtools view -b -o example.region.bam example.sorted.bam 20:1-50000
```

```python
#Alignment Statistics (Flagstat)
print("\nAlignment Statistics:")
!samtools flagstat example.sorted.bam

print("\n Post-alignment analysis completed succesfully!")
```
